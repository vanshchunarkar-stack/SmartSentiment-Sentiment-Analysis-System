# SmartSentiment – Sentiment Analysis System Using NLP

This notebook demonstrates the complete sentiment analysis pipeline: dataset loading, preprocessing, feature extraction, model training, and evaluation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import sys
sys.path.append('../source_code')
from preprocess import clean_text
```

## 1. Load Dataset

```python
df = pd.read_csv('../dataset/sentiment_dataset.csv')
df.head()
```

## 2. Text Preprocessing

```python
df['cleaned_text'] = df['text'].apply(clean_text)
df[['text', 'cleaned_text', 'sentiment']].head()
```

## 3. Feature Extraction & Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(df['cleaned_text'], df['sentiment'], test_size=0.2, random_state=42)
vectorizer = TfidfVectorizer(max_features=5000)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

## 4. Model Training

```python
model = LogisticRegression(random_state=42, multi_class='ovr')
model.fit(X_train_vec, y_train)
```

## 5. Evaluation

```python
y_pred = model.predict(X_test_vec)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred, zero_division=0))
```

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=model.classes_, yticklabels=model.classes_)
plt.title('Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

## 6. Sample Prediction

```python
sample_texts = ['I absolutely love this product.', 'This is the worst experience I have ever had.']
sample_cleaned = [clean_text(t) for t in sample_texts]
sample_vec = vectorizer.transform(sample_cleaned)
predictions = model.predict(sample_vec)
for t, p in zip(sample_texts, predictions):
    print(f'Text: {t}\nPredicted Sentiment: {p.upper()}\n')
```
